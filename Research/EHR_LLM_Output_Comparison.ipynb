## 1. 원본 데이터 확인

```python
import pandas as pd

# 데이터 경로 설정
file_path = "/content/drive/MyDrive/DILAB/MARS/mimic-iv-note_2.2/files/note/discharge.csv"

# CSV 파일 불러오기
print("데이터를 불러오는 중입니다...")
df = pd.read_csv(file_path)

# 깃허브 결과물에 있는 3명의 환자 ID (subject_id) 리스트
# target_patients = [10000032, 10001217, 10001725]
target_patients = [10001217]

# 해당 환자들의 데이터만 필터링
target_df = df[df['subject_id'].isin(target_patients)]

# 결과 확인 및 출력
print(f"총 {len(target_df)}건의 원본 기록을 찾았습니다.\n")

for index, row in target_df.iterrows():
    print("="*50)
    print(f"환자 ID (subject_id): {row['subject_id']}")
    print(f"입원 ID (hadm_id): {row['hadm_id']}")
    print(f"기록 유형 (note_type): {row['note_type']}")
    print("-" * 50)
    print(str(row['text'])+ "...\n")
```

## 2. 파인튜닝 데이터 확인

```python
import pandas as pd

path = "/content/drive/MyDrive/DILAB/Models/Qwen2-7B-Instruct/results/original_1.4/original_1_4_results.parquet"

df = pd.read_parquet(path)

target_patient_id = 10000032

target_df = df[df['subject_id'] == target_patient_id]

print(f"총 {len(target_df)}건의 해당 환자 기록을 찾았습니다\n")

for idx, row in target_df.iterrows():
  print(f"\n======= 인덱스 {idx} (환자 10000032) =======")

  for i, col in enumerate(target_df.columns, start=1):
    value = row[col]
    print(f"{i}. {col}\n{value}\n")
```
