```python
from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline


def run_ner_demo(model_name: str, texts: list, description: str = ""):
    """주어진 모델로 NER pipeline을 실행하고 결과를 보기 좋게 출력"""
    print(f"\n{'=' * 78}")
    print(f"Model: {model_name}")
    if description:
        print(f"Desc : {description}")
    print(f"{'=' * 78}")

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForTokenClassification.from_pretrained(model_name)
    ner = pipeline(
        "ner",
        model=model,
        tokenizer=tokenizer,
        aggregation_strategy="simple",  # subword들을 word 단위로 묶어줌
    )

    for text in texts:
        print(f"\n[INPUT] {text}")
        entities = ner(text)
        if not entities:
            print("   (no entities detected)")
            continue
        for ent in entities:
            print(
                f"   - {ent['word']:30s} | "
                f"{ent['entity_group']:15s} | "
                f"score={ent['score']:.3f}"
            )


if __name__ == "__main__":
    biomedical_texts = [
        "BRCA1 mutations significantly increase the risk of breast cancer and ovarian cancer.",
        "The patient was diagnosed with type 2 diabetes mellitus and prescribed metformin 500mg.",
        "Aspirin reduces inflammation and is used for acute myocardial infarction.",
        "TP53 is a tumor suppressor gene mutated in many cancers including lung adenocarcinoma.",
        "COVID-19, caused by SARS-CoV-2, has been treated with remdesivir and dexamethasone.",
    ]

    # 1) BioBERT - Disease NER (NCBI Disease로 fine-tuning)
    run_ner_demo(
        "alvaroalon2/biobert_diseases_ner",
        biomedical_texts,
        description="BioBERT fine-tuned for Disease NER (NCBI Disease)",
    )

    # 2) BioBERT - Chemical/Drug NER
    run_ner_demo(
        "alvaroalon2/biobert_chemical_ner",
        biomedical_texts,
        description="BioBERT fine-tuned for Chemical NER (BC5CDR-Chem)",
    )

    # 3) BioBERT - Gene/Protein NER
    run_ner_demo(
        "alvaroalon2/biobert_genetic_ner",
        biomedical_texts,
        description="BioBERT fine-tuned for Genetic/Protein NER (BC2GM)",
    )

    # 4) (참고) BioBERT는 아니지만 multi-type 비교용 - 84개 entity type
    run_ner_demo(
        "d4data/biomedical-ner-all",
        biomedical_texts,
        description="[참고] BERT-based biomedical NER, 84 entity types",
    )
```
